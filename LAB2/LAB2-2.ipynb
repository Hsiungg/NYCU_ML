## Import necessary libraries

```python
import numpy as np
import math
```

## Read input file

```python
def read_txt(input_file):
    with open(input_file, 'r') as f:
        lines = f.readlines()
    return [line.strip().split() for line in lines]
```

## Beta and Binomial

```python
class online_beta_binomial:
    def __init__(self, input_array, a, b):
        self.a, self.b = a, b
        self.case_length = len(input_array)
        self.int_arrays = [np.array(list(map(int, row[0]))) for row in input_array]
        
    def Binomial_likelihood(self, theta, n, x):
        return math.comb(n, x) * theta**x * (1-theta)**(n-x)
    
    def iterate_cases(self):
        for i, case in enumerate(self.int_arrays):
            binary_string = ''.join(map(str, case))
            print(f"case {i+1}: {binary_string}")
            x = np.sum(case)
            n = case.shape[0]
            theta = x / n 
            likelihood = self.Binomial_likelihood(theta, n, x)
            self.output(x, n, likelihood)
            self.update(x, n)
    def output(self, x, n, likelihood):
        print(f'Likelihood: {likelihood}')
        print(f'beta prior:     a = {self.a} b = {self.b}')
        print(f'Beta posterior: a = {self.a + x} b = {self.b + n - x}', end='\n\n' )
        
    def update(self, x, n):
        self.a += x
        self.b += (n - x)
        
```

## Experiment

```python
input = read_txt('testfile.txt')
case_1 = online_beta_binomial(input, 0, 0)
case_1.iterate_cases()
```

```python
case_2 = online_beta_binomial(input, 10, 1)
case_2.iterate_cases()
```
